```python
import cv2
import numpy as np
import glob
from os.path import isfile, join
import subprocess
import os
from IPython.display import clear_output

directory = '/content/Real-ESRGAN/videos'
upload_folder = '/content/Real-ESRGAN/upload'
result_folder = '/content/Real-ESRGAN/results'
video_result_folder = '/content/Real-ESRGAN/results_videos'
video_mp4_result_folder = '/content/Real-ESRGAN/results_mp4_videos'

zee = 0

os.makedirs(upload_folder, exist_ok=True)
os.makedirs(result_folder, exist_ok=True)
os.makedirs(video_result_folder, exist_ok=True)
os.makedirs(video_mp4_result_folder, exist_ok=True)

for f in os.listdir(upload_folder):
    f_path = os.path.join(upload_folder, f)
    if os.path.isfile(f_path):
        os.remove(f_path)

for f in os.listdir(result_folder):
    f_path = os.path.join(result_folder, f)
    if os.path.isfile(f_path):
        os.remove(f_path)

for f in os.listdir(video_result_folder):
    f_path = os.path.join(video_result_folder, f)
    if os.path.isfile(f_path):
        os.remove(f_path)

for f in os.listdir(video_mp4_result_folder):
    f_path = os.path.join(video_mp4_result_folder, f)
    if os.path.isfile(f_path):
        os.remove(f_path)

def convert_frames_to_video(pathIn, pathOut, fps):
    frame_array = []
    files = [f for f in os.listdir(pathIn) if isfile(join(pathIn, f))]
    files.sort(key = lambda x: int(x[5:-4]))
    size2 = (0,0)

    for i in range(len(files)):
        filename = pathIn + files[i]
        img = cv2.imread(filename)
        height, width, layers = img.shape
        size = (width, height)
        size2 = size
        frame_array.append(img)

    out = cv2.VideoWriter(pathOut, cv2.VideoWriter_fourcc(*'DIVX'), fps, size2)
    for i in range(len(frame_array)):
        out.write(frame_array[i])
    out.release()

for filename in os.listdir(directory):
    f = os.path.join(directory, filename)
    if os.path.isfile(f):
        print("PROCESSING :" + str(f) + "\n")

        # Save temp video safely outside the frames folder
        smoothed_video = '/content/Real-ESRGAN/smoothed_video_temp.mp4'
        subprocess.call([
            'ffmpeg', '-y', '-i', f,
            '-vf', 'deflicker=size=10:mode=pm',
            '-c:a', 'copy', smoothed_video
        ])

        cam = cv2.VideoCapture(smoothed_video)

        # Dynamically extract correct FPS for perfect audio sync
        video_fps = cam.get(cv2.CAP_PROP_FPS)
        if video_fps == 0 or np.isnan(video_fps):
            video_fps = 25.0

        currentframe = 0
        while(True):
            ret, frame = cam.read()
            if ret:
                name = os.path.join(upload_folder, 'frame' + str(currentframe) + '.jpg')
                cv2.imwrite(name, frame)
                currentframe += 1
            else:
                break
        cam.release()

        !python inference_realesrgan.py -n RealESRGAN_x4plus -i /content/Real-ESRGAN/upload --outscale 2 --tile 256

        for file_name in os.listdir(upload_folder):
            file_path = os.path.join(upload_folder, file_name)
            if os.path.isfile(file_path):
                os.remove(file_path)

        paths = (os.path.join(root, fn)
                 for root, _, filenames in os.walk('/content/Real-ESRGAN/results')
                 for fn in filenames)
        for path in paths:
            newname = path.replace('_out', '')
            if newname != path:
                os.rename(path, newname)

        pathIn = '/content/Real-ESRGAN/results/'
        zee = zee + 1
        fName = "video" + str(zee)
        filenameVid = f"{fName}.avi"
        pathOut = os.path.join(video_result_folder, filenameVid)

        convert_frames_to_video(pathIn, pathOut, video_fps)

        # Fixed target folder cleanup so frames don't bleed into the next video
        for file_name in os.listdir(result_folder):
            file_path = os.path.join(result_folder, file_name)
            if os.path.isfile(file_path):
                os.remove(file_path)

        src = video_result_folder
        dst = video_mp4_result_folder

        for root, dirs, filenames in os.walk(src, topdown=False):
            for fn in filenames:
                try:
                    _format = ''
                    if ".flv" in fn.lower(): _format = ".flv"
                    if ".mp4" in fn.lower(): _format = ".mp4"
                    if ".avi" in fn.lower(): _format = ".avi"
                    if ".mov" in fn.lower(): _format = ".mov"

                    inputfile = os.path.join(root, fn)
                    outputfile = os.path.join(dst, fn.lower().replace(_format, ".mp4"))

                    subprocess.call([
                        'ffmpeg', '-y', '-i', inputfile, '-i', f,
                        '-c:v', 'libx264', '-c:a', 'aac',
                        '-map', '0:v:0', '-map', '1:a:0',
                        '-shortest', outputfile
                    ])
                except:
                    print("An exception occurred")

        clear_output(wait=True)

        for file_name in os.listdir(video_result_folder):
            file_path = os.path.join(video_result_folder, file_name)
            if os.path.isfile(file_path):
                os.remove(file_path)

        if os.path.exists(smoothed_video):
            os.remove(smoothed_video)
```
